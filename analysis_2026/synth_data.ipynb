```python
import numpy as np
import pandas as pd
import pickle
import glob
import os
import random
```

```python
class EEG_data:
    def __init__(self, label, time, eeg):
        self.time = time
        self.eeg = eeg
        self.label = label
```

```python
data_dir = 'data_synth'
```

```python
files = glob.glob(os.path.join(data_dir, '*.pickle'))
```

```python
files
```

```python
for file in files:
    data = pd.read_pickle(file)
    for d in data:
        if hasattr(d, 'eeg'):
            np.random.shuffle(d.eeg)
        if hasattr(d, 'f_eeg'):
            np.random.shuffle(d.f_eeg)
    with open(file, 'wb') as f:
        pickle.dump(data, f)
```
